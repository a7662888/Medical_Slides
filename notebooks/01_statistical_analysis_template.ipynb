# Statistical Analysis Template

Use this notebook for descriptive tables, group comparisons, missingness review, and manuscript-ready summary outputs.

```python
from pathlib import Path

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from dementia_ai_research.preprocessing import clean_dataset

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
RESULTS_DIR = PROJECT_ROOT / 'results'
TABLES_DIR = RESULTS_DIR / 'tables'
FIGURES_DIR = RESULTS_DIR / 'figures'
TABLES_DIR.mkdir(parents=True, exist_ok=True)
FIGURES_DIR.mkdir(parents=True, exist_ok=True)
```

## Load De-identified Dataset

Keep raw or identifiable data outside GitHub. Use a local path or approved secure storage.

```python
# Example only. Replace with an approved local de-identified dataset path.
# df_raw = pd.read_csv(PROJECT_ROOT / 'restricted_data' / 'analysis_dataset.csv')
# df = clean_dataset(df_raw)
df = pd.DataFrame()
```

## Data Quality Review

```python
if not df.empty:
    display(df.filter(like='qc_').mean().rename('proportion_flagged'))
    display(df.groupby('diagnosis_group', dropna=False).size().rename('n'))
```

## Descriptive Table

```python
continuous_vars = ['age', 'education_years', 'mmse', 'moca', 'cdr_global']

if not df.empty:
    summary = df.groupby('diagnosis_group')[continuous_vars].agg(['count', 'mean', 'std', 'median'])
    display(summary)
    summary.to_csv(TABLES_DIR / 'descriptive_summary_by_group.csv')
```

## Figure Template

```python
if not df.empty and {'diagnosis_group', 'moca'}.issubset(df.columns):
    ax = sns.boxplot(data=df, x='diagnosis_group', y='moca')
    ax.set_xlabel('Diagnosis group')
    ax.set_ylabel('MoCA')
    plt.xticks(rotation=30, ha='right')
    plt.tight_layout()
    plt.savefig(FIGURES_DIR / 'moca_by_diagnosis_group.png', dpi=300)
    plt.show()
```
